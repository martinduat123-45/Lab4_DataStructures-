```python
festival = {"Ben&Ben", "SB19", "Bini", "Eraserheads", FAVORITE_ARTIST, "Zild", f"Indie Artist {CONTROL_NUM}"}

user_a = {"Ben&Ben", "Bini", "Maki", "Dionela", FAVORITE_ARTIST}
user_b = {"SB19", "Eraserheads", "Zild", f"Indie Artist {CONTROL_NUM}", "Parokya ni Edgar"}

festival.update({
    "Moira", "December Avenue", "IV of Spades", "Unique"
})

def jaccard(set1, set2):
    intersection = set1 & set2
    union = set1 | set2
    return len(intersection), len(union), (len(intersection) / len(union)) * 100

int_a, union_a, jacc_a = jaccard(user_a, festival)
int_b, union_b, jacc_b = jaccard(user_b, festival)

dealbreakers = user_a - festival

print("Total Festival Artists:", len(festival))
print("User A Intersection:", int_a)
print("User A Union:", union_a)
print("User A Similarity (%):", jacc_a)
print("User B Similarity (%):", jacc_b)
print("User A Dealbreakers:", dealbreakers)
```

```python
from collections import deque, Counter

dq = deque(maxlen=5)

burst = [
    (CONTROL_NUM, "WEATHER"),
    (CONTROL_NUM + 1, "TRAFFIC"),
    (CONTROL_NUM + 2, "WEATHER"),
    (CONTROL_NUM + 3, "WEATHER"),
    (CONTROL_NUM + 4, LOCAL_HAZARD),
    (CONTROL_NUM + 5, "WEATHER"),
    (CONTROL_NUM + 6, "TRAFFIC")
]

start = CONTROL_NUM + 7
burst.extend([
    (start, "WEATHER"),
    (start+1, "TRAFFIC"),
    (start+2, LOCAL_HAZARD),
    (start+3, "WEATHER"),
    (start+4, "TRAFFIC"),
    (start+5, LOCAL_HAZARD)
])

weather_at_check = None
hazard_check = None

for ts, category in burst:
    current_categories = [c for _, c in dq]
    count = Counter(current_categories)
    
    if count[category] >= 2:
        if ts == CONTROL_NUM + 3:
            weather_at_check = "N"
        if category == LOCAL_HAZARD and hazard_check is None:
            hazard_check = "N"
        continue
    else:
        dq.append((ts, category))
        if ts == CONTROL_NUM + 3:
            weather_at_check = "Y"
        if category == LOCAL_HAZARD and hazard_check is None:
            hazard_check = "Y"

timestamp_sum = sum(ts for ts, _ in dq)

print("LOCAL_HAZARD:", LOCAL_HAZARD)
print("Total Burst Size:", len(burst))
print("Did WEATHER at", CONTROL_NUM+3, "append?:", weather_at_check)
print("Did LOCAL_HAZARD append?:", hazard_check)
print("Final Deque Length:", len(dq))
print("Sum of Timestamps:", timestamp_sum)
print("Final Deque:", dq)
```
